# Quiz 5 - Clustering

```python
import pandas as pd
import numpy as np
import sys

sys.path.append("..")

from AI_algorithms.ml.unsupervised import KMeans
from AI_algorithms.utils.data import parse_table
```

```python
data = '''
id	x0	x1	x2
0	0	0	1
1	1	0	0
2	1	1	1
3	1	1	0
4	0	0	0
5	0	1	1
'''
centroids = np.array([
    [0.79, 0.21, 0.38],
    [0.81, 0.05, 0.06],
    [0.97, 0.42, 0.44]
])
df = parse_table(data, index="index", columns=True).astype(float)
df.head()
```

```python
model = KMeans(n_clusters=3, max_iter=1, tol=1e-4)
model.fit(df.values, centroids = centroids, verbose=True)
```

```python
y = model.predict(df.values)
df_ = df.copy()
df_["cluster"] = y
df_.head()
```
